```python
#f-string report
model_name = "latency"
accuracy = 0.956
epochs = 10
print(f"Model {model_name} achieved {accuracy:.1%} accuracy after {epochs} epochs.")
```

```python
#conditionals 
accuracy = 0.92
latency = 120
#latency and accuracy all goo
if accuracy >= 0.90 and latency <= 100:
    decision = "ship it"

#latency too slow 
elif accuracy >= 0.90 and latency > 100:
    decision = "too slow"

# Accuracy not enough
else:
    decision = "keep training"
print(decision)
    
```

```python
#loop using enumerate
scores = [0.82, 0.87, 0.91, 0.95]

for epoch, score in enumerate(scores, start=1):
    print(f"epoch {epoch}: {score}")
```

```python
#doing collections
# List for ordered and  changed model.
scores_list = [0.82, 0.87, 0.91, 0.95]
print("List:", scores_list)

# Tuple for model information that should not be changed.
model_info = ("Neural Network", 4)
print("Tuple:", model_info)

# Dict for value  and meaningful key.
model = {"accuracy": 0.95, "latency": 120}
print("Dict:", model)

# Set for duplicate removal.
labels = {"AI", "ML", "AI", "Python"}
print("Set:", labels)
```

```python
#function to normarise
values = [12, 18, 25, 40]
minimum = 12
maximum = 40

def normalise(values, minimum, maximum):
    return [(value - minimum) / (maximum - minimum) for value in values]


print(normalise(values, 12, 40))
```
